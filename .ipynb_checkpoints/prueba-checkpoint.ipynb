```python
# Tratamiento de datos
# ==============================================================================
import pandas as pd
import numpy as np

# Gráficos
# ==============================================================================
import matplotlib.pyplot as plt
import matplotlib.font_manager
from matplotlib import style
style.use('ggplot') or plt.style.use('ggplot')

# Preprocesado y modelado
# ==============================================================================
from sklearn.decomposition import PCA
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import scale
```

```python
# Cargamos los datos
data = pd.read_csv('GSE92914_raw_counts.csv', sep=r'\s+', engine='python', index_col=0)

data.head(4)
```

# Exploración inicial

Identificamos el valor promedio y dispersión de las variables

```python
print('----------------------')
print('Media de cada variable')
print('----------------------')
data.mean(axis=0)
```

```python
print('-------------------------')
print('Varianza de cada variable')
print('-------------------------')
data.var(axis=0)
```

# Modelo PCA


```python
# Entrenamiento modelo PCA con escalado de los datos
# ==============================================================================
pca_pipe = make_pipeline(StandardScaler(), PCA())
pca_pipe.fit(data.transpose())
```

```python
import seaborn as sns
pca_result = pca_pipe.transform(data.transpose())

# 2. Meterlo en un DataFrame
pca_df = pd.DataFrame(
    pca_result[:, :2],  # Solo PC1 y PC2
    columns=['PC1', 'PC2'],
    index=data.columns  # El índice serán las muestras
)

# 3. Añadir la información del grupo
# Supongamos que tienes una lista llamada "group" que indica el grupo de cada muestra
# Ejemplo:
group = ["Normal", "Normal", "Normal", "Tumor", "Tumor", "Tumor", "Metastasis", "Metastasis", "Metastasis", "Normal", "Tumor", "Metastasis"]

pca_df['Group'] = group

# 4. (Opcional) Si quieres también poner los nombres de pacientes
# pca_df['Patient'] = patient_list

# 5. Graficar
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=pca_df, 
    x='PC1', y='PC2', 
    hue='Group', 
    style='Group', 
    s=100
)
```

```python
# Se extrae el modelo entrenado del pipeline
modelo_pca = pca_pipe.named_steps['pca']
```

## Interpretación

```python
# Se combierte el array a dataframe para añadir nombres a los ejes.
pd.DataFrame(
    data    = modelo_pca.components_,
    columns = data.transpose().columns,
    index   = ['PC1', 'PC2', 'PC3', 'PC4', 'PC5','PC6','PC7','PC8','PC9','PC10','PC11','PC12' ]
)
```

Analizamos la influencia de cada variable en cada componente, visualmente mediante un gráfico del tipo heatmap:

```python
# Heatmap componentes
# ==============================================================================
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(4, 2))
componentes = modelo_pca.components_
plt.imshow(componentes.T, cmap='viridis', aspect='auto')
plt.yticks(range(len(data.transpose().columns)), data.columns)
plt.xticks(range(len(data.transpose().columns)), np.arange(modelo_pca.n_components_) + 1)
plt.grid(False)
plt.colorbar();
```

```python
# Porcentaje de varianza explicada por cada componente
# ==============================================================================
print('----------------------------------------------------')
print('Porcentaje de varianza explicada por cada componente')
print('----------------------------------------------------')
print(modelo_pca.explained_variance_ratio_)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(6, 4))
ax.bar(
    x      = np.arange(modelo_pca.n_components_) + 1,
    height = modelo_pca.explained_variance_ratio_
)

for x, y in zip(np.arange(len(data.columns)) + 1, modelo_pca.explained_variance_ratio_):
    label = round(y, 2)
    ax.annotate(
        label,
        (x,y),
        textcoords="offset points",
        xytext=(0,10),
        ha='center'
    )

ax.set_xticks(np.arange(modelo_pca.n_components_) + 1)
ax.set_ylim(0, 1.1)
ax.set_title('Porcentaje de varianza explicada por cada componente')
ax.set_xlabel('Componente principal')
ax.set_ylabel('Por. varianza explicada');
```

```python
# Porcentaje de varianza explicada acumulada
# ==============================================================================
prop_varianza_acum = modelo_pca.explained_variance_ratio_.cumsum()
print('------------------------------------------')
print('Porcentaje de varianza explicada acumulada')
print('------------------------------------------')
print(prop_varianza_acum)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(6, 4))
ax.plot(
    np.arange(len(data.columns)) + 1,
    prop_varianza_acum,
    marker = 'o'
)

for x, y in zip(np.arange(len(data.columns)) + 1, prop_varianza_acum):
    label = round(y, 2)
    ax.annotate(
        label,
        (x,y),
        textcoords="offset points",
        xytext=(0,10),
        ha='center'
    )
    
ax.set_ylim(0, 1.1)
ax.set_xticks(np.arange(modelo_pca.n_components_) + 1)
ax.set_title('Porcentaje de varianza explicada acumulada')
ax.set_xlabel('Componente principal')
ax.set_ylabel('Por. varianza acumulada');
```

Transformación de los datos y reducción de la dimensionalidad

```python
# Proyección de las observaciones de entrenamiento
# ==============================================================================
proyecciones = pca_pipe.transform(X=data)
proyecciones = proyecciones[:,:3]
proyecciones = pd.DataFrame(
    proyecciones,
    columns = ['PC1', 'PC2', 'PC3'],
    index   = data.index
)
proyecciones.head()
```

```python
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(
    proyecciones['PC1'], 
    proyecciones['PC2'], 
    proyecciones['PC3'],
    alpha=0.6
)

ax.set_xlabel('PC1')
ax.set_ylabel('PC2')
ax.set_zlabel('PC3')
ax.set_title('Proyección de genes en PC1-PC2-PC3')
plt.show()
```

# Custering

Hagamos clustering con las 12 variables, como si no hubiéramos hecho PCA, aunque sí vamos a escalar los datos

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Primero, escalamos los datos
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data[['GSM2440320', 'GSM2440321', 'GSM2440322', 'GSM2440323', 'GSM2440324', 
                                        'GSM2440325', 'GSM2440326', 'GSM2440327', 'GSM2440328', 'GSM2440329', 
                                        'GSM2440330', 'GSM2440331']])

# Aplicamos KMeans sobre los datos escalados
kmeans = KMeans(n_clusters=3, random_state=0)
clusters = kmeans.fit_predict(data_scaled)

# Guardamos la asignación de clusters en el DataFrame
data['cluster'] = clusters

# Ver las primeras filas para comprobar
data.head()
```

```python
from sklearn.decomposition import PCA
import plotly.express as px

data_scaled = pd.DataFrame(data_scaled, columns=['GSM2440320', 'GSM2440321', 'GSM2440322', 'GSM2440323', 'GSM2440324', 
                                                'GSM2440325', 'GSM2440326', 'GSM2440327', 'GSM2440328', 'GSM2440329', 
                                                'GSM2440330', 'GSM2440331'])
# Reducir la dimensionalidad de las 12 variables a 3 con PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(data_scaled[['GSM2440320', 'GSM2440321', 'GSM2440322', 'GSM2440323', 'GSM2440324', 
                                    'GSM2440325', 'GSM2440326', 'GSM2440327', 'GSM2440328', 'GSM2440329', 
                                    'GSM2440330', 'GSM2440331']])

# Añadir el resultado del PCA y los clústeres a la DataFrame
data_scaled['PC1'] = pca_result[:, 0]
data_scaled['PC2'] = pca_result[:, 1]
#data_scaled['PC3'] = pca_result[:, 2]

# Ya tienes el clustering
data_scaled['cluster'] = clusters

# Ahora el gráfico 3D
fig = px.scatter_2d(
    data_scaled,
    x='PC1', y='PC2',
    color='cluster',
    opacity=0.7
)

fig.update_layout(title='PCA 2D interactivo con Clustering KMeans')
fig.show()
```

Ahora que hemos reducido la dimensionalidad, podemos hacer un clustering más fino. Aplicaremos un algoritmo de clustering no jerárquico: el método Kmeans.

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=0)
clusters = kmeans.fit_predict(proyecciones[['PC1', 'PC2', 'PC3']])

proyecciones['cluster'] = clusters  # Guardamos el cluster al que pertenece cada gen
```

```python
import plotly.express as px

fig = px.scatter_3d(
    proyecciones,
    x='PC1', y='PC2', z='PC3',
    color='cluster',
    opacity=0.7
)

fig.update_layout(title='PCA 3D interactivo con Plotly')
fig.show()
```
